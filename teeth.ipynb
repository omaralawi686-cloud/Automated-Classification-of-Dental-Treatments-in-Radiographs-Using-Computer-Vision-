```python
import pandas as pd
import os
```

```python
csv_path = r"C:\Users\landa\Desktop\annotations.train.csv"
base_dir = r"C:\Users\landa\Desktop\train images"
output_dir = os.path.join(base_dir,"labels")
os.makedirs(output_dir , exist_ok = True)
```

```python
image_folder = r"C:\Users\landa\Desktop\train images"
all_images = [f for f in os.listdir(image_folder) if f.endswith((".jpg" , ".jpeg" , ".png"))]
print (all_images)
```

```python
df = pd.read_csv(csv_path)
```

```python
df['class'] = df['class'].str.strip().str.capitalize()

annotated_images = df['filename'].unique()
annotated_images = [os.path.basename(f) for f in annotated_images]

classes= sorted(df['class'].unique())
class_map = {cls : idx for idx , cls in enumerate(classes)}
print (" class map :" ,class_map)
```

```python
def convert_to_yolo(row,width,height,class_map):
    class_id = class_map[row['class']]
    x_center = (row['xmin'] + row['xmax']) / 2 / width
    y_center = (row ['ymin'] + row['ymax']) / 2 / height
    box_width = (row ['xmax'] - row ['xmin']) / width
    box_height = (row ['ymax'] - row ['ymin']) / height

    return f"{class_id} {x_center} {y_center} {box_width} {box_height}"
    
    
```

```python
grouped = df.groupby('filename')
for filename , group in grouped:
    output_lines =[]
    width = group ['width'].iloc[0] 
    height = group ['height'].iloc[0]


    for _, row in group.iterrows():
        yolo_line = convert_to_yolo(row,width,height,class_map)
        output_lines.append(yolo_line)
        output_filepath = os.path.join(output_dir ,f"{filename.split('.')[0]}.txt")
```

```python
with open(output_filepath, "w") as f:
    f.write("\n".join(output_lines))
```
